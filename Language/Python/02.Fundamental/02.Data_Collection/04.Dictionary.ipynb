### What is Dictionary?

Dictionaries are
- Mutable
- insertion-ordered collections of key-value pairs.
- Key must be "unique" and "immutable".
- Value can be any type.

Use Case:
- Insertion ordered.
- Fast lookups by key.
- Ideal for configuration data, JSON-like structure and lookups.
- Mutable: add, remove or change key-value pair.

```python
# Define and print dictionary
my_dict = {'a':1, 'b':2, 'c':3}
print(my_dict)

# print the length, only keys, only values, items
print(f"Length: {len(my_dict)}")
print(f"Keys: {my_dict.keys()}")
print(f"Values: {my_dict.values()}")
print(f"Values: {my_dict.items()}") # Returns a collection of tuples

# Prove items return collection of tuples
for i in my_dict.items():
    print(type(i))

# retrive key & value from items
for k,v in my_dict.items():
    print(f"- {k}: {v}")

# check a key is part of dictionary
print("b" in my_dict)
print("d" in my_dict)

# check a value is part of dictionary
print(1 in my_dict)
print(1 in set(my_dict.values()))
```

```python
print(my_dict['b'])
# print(my_dict['d']) # KeyError
print(my_dict.get('b'))
print(my_dict.get('d'))
# Return defaukt value if key not exist.
print(my_dict.get('d', -1))
```

```python
my_dict = {'a':1, 'b':2, 'c':3}
# Add item to a dictionary using inbuild functions
my_dict.setdefault('d', 4)
print(my_dict)
# Remove last item from the dictionary.
removed = my_dict.popitem()
print(removed)
print(my_dict)
# Remove an item and print the value
removed = my_dict.pop('b')
print(removed)
print(my_dict)
```

```python
default_tags = {
    "Environment": "Prod",
    "Owner": "Finance",
    "CostCode": 10000,
}
custom_tag = {
    "team": "techops",
    "CostCode": 12345,
}
# Merge and create new dictionary without modify the original dictionary.
merged_tag = default_tags | custom_tag
print(merged_tag)
# Merge and modify the orginal dictionary
default_tags.update(custom_tag)
print(default_tags)
```

```python
# Create a new dictionary based on the set of keys.
new_dict = dict.fromkeys(["one", "two", "one"], 0)
print(new_dict)
# Clear all the items in dictionary.
new_dict.clear()
print(new_dict)
```

```python
tags = {
    "Environment": "Prod",
    "Owner": "Finance",
    "CostCode": 10000,
}
print(tags)
# Adding items to dictionary
tags["team"] = "techops"
# Updating items in dictionary.
tags["CostCode"] = 12345
print(tags)
```
