### What is tuples
Tuples are 
- ordered
- immutable 
- Allow duplicates
- Defined within parenthesis()

Use case: Data is fixed and positional, with multiple ordered values representing a higher construct.
- eg) redis_cfg = ('10.0.5.1', 6379)
- (host, port) pair
- RGB color
- Multiple-return values: return(stdout, stderr, rc) and callers unpack it.

```python
host_port = ("127.0.0.1", 3000)
print(host_port)
red_rgb = (255, 0, 0)
print(type(red_rgb))
tuple_single = ("a")
print(type(tuple_single))
tuple_single = ("a",)
print(type(tuple_single))
```

### Accessing tuples

```python
print(host_port[0])
print(type(host_port[0]))

print(red_rgb[-2:])
print(type(red_rgb[-2:]))

# TypeError is raised as tuple is immutable.
# red_rgb[1] = 255 
```

```python
service_endpoint = ('auth-server.dev.local', 8080)
print(f"Hostname: {service_endpoint[0]}, Port: {service_endpoint[1]}")
# service_endpoint[1] = 3000
```
