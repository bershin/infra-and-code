### What is Sets
Sets are 
- unordered
- mutable
    - The item of the set must be immutable.(set of tuples, NOT list or sets)
- unique items only(no duplicates). 
- Represent by flower bracker {} or set([])

Usecases: Focus is on deduplicating the data or performing membership checks.
- eg) allowed_ports = {22, 80, 443}
- Deduplicating values in multiple occasions.(IPs extract from a series of access logs)
- Allowed-ports whitelist.

```python
unique_num_ports = set([80, 443, 22, 8080, 80, 443])
print(unique_num_ports)
native_unique_num_port = {80, 443, 22, 8080, 80, 443}
native_unique_str_host = {"web01", "web02", "web03", "web02"}
print(native_unique_str_host)

## Sets cannot have mutable elements
# sets_of_list = set([[1,2],[3,4],[5,6]])
# nat_set_list = {[1,2],[3,4],[5,6]}
# sets_of_sets = set([{1,2},{3,4},{5,6}])
# nat_set_sets = {{1,2},{3,4},{5,6}}
sets_of_tuples = set([(1,2),(3,4)])
print(sets_of_tuples)
nat_sets_tuples = {(1,2),(3,4)}
print(nat_sets_tuples)
```

### Accessing sets

```python
print(22 in unique_num_ports)
print(22 in native_unique_str_host)

print((1,2) in sets_of_tuples)
print((2,1) in sets_of_tuples)
```

### Modifying - NO order

```python
ports = set([80, 443, 22, 8080, 80, 443])
print(ports)
ports.add(3000)
print(ports)
ports.remove(22)
print(ports)
# ports.remove(22)
# Safer way to remove
ports.discard(22)
print(ports)
```

### Set operations

```python
developers = set(["alice", "bob", "charlie"])
admins = set(['alice', 'david'])
print("alice" in developers)
print("alice" in admins)

print(developers.union(admins))
print(developers | admins)

print(developers.intersection(admins))
print(developers & admins)

print(developers.difference(admins))
print(developers - admins)

print(developers)
print(admins)

print(developers.intersection_update(admins))
print(developers)
print(admins)
```
