### What is List?

List are 
- ordered
- mutable 
- Allow duplicates
- Defined with square brackets [].

Use Case: When you need a editable, positional aware sequence.
- eg) servers = ['web01', 'db01', 'web01']
- Ordered deployment steps
- Log lines or file reads
- Task queues

### Accessing list: Index & Slice

```python
# List of same type & mixed type, allowed in python and print.
server_names = ["web01", "web02", "web03"]
mixed_list = ["config.yaml", 8080, True]

for item in mixed_list:
    print(item, type(item))

# Print first element, last element, last before element, index out of range error.
print(server_names[0])
print(server_names[-1])
print(server_names[-2])
# print(server_names[5])

# Print first two element in list | last two element ( 2-ways)
print(server_names[0:2])
print(server_names[:2])
print(server_names[-2:])
print(server_names[1:])

# Print alternate element in the list
print(server_names[::2])

# Prove slicing do not alter the original list
print(server_names)
```

### Modify list content: insert, remove, pop, append

```python
ports = [22, 25, 80, 443, 8080]
print(ports)
ports.append(5000)
print(ports)
# ports.remove? --> get help
ports.remove(80)
print(ports)
ports.insert(1, 3000)
print(ports)
ports.pop()
print(ports)
ports.pop(2)
print(ports)
```

### Copy argument to function do not mutate

```python
def mutate_list(lst):
    lst.pop()

alpha = ["a", "b", "c"]
mutate_list(alpha)
print(alpha)
```
