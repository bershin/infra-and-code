### Where to use "for" and "while"
For:
- iteration over item in known sequence
- Used: list, tuple, string, dictionary item, range, file lines

while:
- iterate as long as the condition is true.
- Used: when number of iteration isn't known beforehand.

```python
servers = ["web01", "web02", "web03"]
for server in servers:
    print("pinging server", server)

status = "Success"
for char in status:
    print(char)

for num in range(10):
    print(num)
```

```python
connected = False
conection_attempt = 0
max_attempt = 5

while not connected and conection_attempt < max_attempt:
    print(f"Connection attempt {conection_attempt + 1} failed. Retring...")
    if conection_attempt == 3:
        connected = True
    conection_attempt += 1

if not connected:
    print("Failed to connect after maximum attempts.")
```

Break:
- immediately exist the innermost loop.
- Useful: when you found what you need(hit or error)
Continue:
- Skip the rest of the current iteration and move to next.
- USeful: For skipping item that don't meet criteria. 

```python
users = ["guest", "tester01", "admin01", "admin02", "dev01"]
for user in users:
    print(f"Checking user {user}")
    if user.startswith("admin"):
        print(f"Admin user {user} found")
```

```python
users = ["guest", "tester01", "admin01", "admin02", "dev01"]
for user in users:
    print(f"Checking user {user}")
    if user.startswith("admin"):
        print(f"Admin user {user} found")
        break
```

```python
filenames = ["nginx.conf", "app.yaml", "db.yaml", "notes.txt"]
for file in filenames:
    if not file.endswith("yaml"):
        print(f"Skipping file {file}")
        continue
    print(f"Processing file {file}")
```
