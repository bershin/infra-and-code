### Falsy values
- Boolean -> False
- Object -> None
- Numeric type ->
    - int -> 0
    - float -> 0.0
- Empty collections ->
    - list -> []
    - tuples -> ()
    - strings -> ""
    - dict -> {}

```python
users = ["user1", "user2"]
if users:
    print("Users has elements")
else:
    print("Users is empty")

config = {}
if config:
    print("Config has elements")
else:
    print("Config is empty")
```

### When "if" condition executes its block of code.
Only if the conditions is evaluated to Truthy
- Comparison operators: ==, !=, <, <=, >, >=, in
- Combine conditions: and, or, not

```python
server_status = "running"
servers = ["",""]
default_config = {}
error_message = ""

if server_status:
    print("Service is active")

if servers:
    print("Servers are present")

if error_message:
    print("Something went wrong")

if not default_config:
    print("Config is empty")

if "r" in server_status:
    print("Service is active")

if "e" not in server_status:
    print("Service is inactive")
```

```python
cpu_usage = 85.0

if cpu_usage > 90.0:
    print("Alert: High CPU usage.")
else:
    print("Info: Usage is normal.")
```

```python
http_status = 301
if http_status == 201:
    print("Status OK")
elif http_status == 401:
    print("Resource not found")
elif http_status >= 500:
    print("Server Error (5xx)")
else:
    print("Internal error", http_status)
```

### What is Guard clauses
Handle edge cases at top of the function to avoid deep nesting of "if" condition.

```python
def process_data(data):
    print(f"processing {len(data)} items..")

process_data([])
process_data([1,2,3])
process_data("abc")
# process_data(None) # TypeError None has no len()
```

```python
def process_data(data):
    if not data:
        print("No data provided")
    else:    
        print(f"processing {len(data)} items..")

process_data([])
process_data([1,2,3])
process_data("abc")
process_data(None) # TypeError None has no len()
```

```python
def process_data(data):
    if not data:
        print("No data provided")
    elif not isinstance(data, (list,tuple)):
        print(f"Cannot process {type(data)} only accepts list/tuples")
    else:    
        print(f"processing {len(data)} items..")

process_data([])
process_data([1,2,3])
process_data(("a","b","c"))
process_data("abc")
process_data(None) # TypeError None has no len()
```
