### What is list comprehension
Create a list directly inside [], using loops and conditions.

```python
my_num_list = [2, 4, 3, 7, 5]
double_num_list = []
for num in my_num_list:
    double_num_list.append(num * 2)
print(double_num_list)
```

```python
double_num_list = [num * 2 for num in my_num_list]
print(double_num_list)
```

```python
servers = ["web", "backend", "db"]
uc_server_list = [server.upper() for server in servers]
print(uc_server_list)
```

```python
num_list = [1, 5, 10, 8, 2, 15]
even_num_list = [num for num in num_list if num % 2 == 0]
print(even_num_list)
```

```python
filenames = ["PROD.YML", "settings.json", "app.yaml", "DEV.YML", "conftest.py"]
yaml_config = [f.lower() for f in filenames if f.lower().endswith(('yml', 'yaml'))]
print(yaml_config)
```

```python
numbers = [1, 2, 3, 2, 4, 1, 3]
unique_squares = {num * num for num in numbers}
print(unique_squares)
```

```python
servers = ["web", "be"]
server_ips = {server: f"192.168.1.{i}" for i, server in enumerate(servers, start=10)}
# server_ips = {server: f"192.168.1.{i+10}" for i, server in enumerate(servers)}
print(server_ips)
```

```python
roles = ['primary-db', 'replica-db', 'cache']
hosts = ['db01.prod', 'db02.prod']
 
server_map = {role: host for role, host in zip(roles, hosts)}
print(server_map)
```

```python
num_list = [1, 5, 10, 8, 2, 15]
even_num_list = ["play_out" if num > 8 else "play_in" for num in num_list]
print(even_num_list)
```

```python
num_list = [1, 5, 10, 8, 2, 15]
even_num_list = ["play_out" if num > 8 else "play_in" for num in num_list if num % 2 == 0]
print(even_num_list)
```

```python
servers = [('sv-1', 'us-east-1'), ('sv-2', 'eu-west-1'), ('sv-3', 'us-west-2')]
server_categories = {
    name: "USA" if region.startswith('us-') else "Europe" for name, region in servers
}
print(server_categories)
```

```python
all_required = {'git', 'docker', 'python'}
environments = {
    'ci-runner': {'git', 'docker'},
    'dev-vm': {'git', 'python', 'vscode'}
}
 
for env, installed in environments.items():
    missing = all_required.difference(installed)
    if missing:
        # print(f"Env '{env}' is missing: {missing}")
        print(f"Env '{env}' is missing: {list(missing)}")
        # print(f"Env '{env}' is missing: {sorted(list(missing))}")
```

### Difference between conditional expression before and after loop
- After loop: Deciside the result value to be stored or not
- Before loop: process on the result value which definitely get stored.
