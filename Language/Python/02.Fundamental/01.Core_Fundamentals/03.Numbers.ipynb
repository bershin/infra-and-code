Arithmetic operation

```python
# Addition
print(2+3)
print(2+0.3)
print(0.4+0.6)
# Substract
print(3-1)
print(2-0.3)
print(0.6-0.4)
# Multiplication
print(3*4)
print(2*0.3)
print(0.6*0.4)
```

Division always evaluated to float

```python
print(16/2)
print(7/2)
```

Floor division, Modulo, Power

```python
# Floor division
print(8//2)
print(7//2)
print(7//2.0)
# Modulo
print(8%2)
print(7%2)
# Power
print(2**3)
```
