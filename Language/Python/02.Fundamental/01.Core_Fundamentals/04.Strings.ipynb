What is String?


String is a ordered, immutable, sequence of character

Ways to create a string, Single & multiline strings.

```python
last_name = 'john'
first_name = "Berchin"
multi_line = """
No intend
    have some intend
"""
print(last_name)
print(first_name)
print(multi_line)
```

Formating output using fstring

```python
math_division = 3/7
print(f"Format Result: {math_division}")
print(f"Inline expression evaluation: {3/7}")
```

Essential string methods

Strip - Removing white spaces

```python
course_title = "       Python for Devops.     "
print(f"Course title {course_title} is good.")
print(f"Course title {course_title.strip()} is good.")
print(f"Course title {course_title.rstrip()} is good.")
print(f"Course title {"       Python for Devops.     ".lstrip()} is good.")
```

Changing case

```python
name = "bershin John"
print(f"My name is {name.upper()}")
print(f"My name is {"Bershin John".lower()}")
```

Start and end

```python
file = "pod.yaml"
print(f"File name starts with po: {file.startswith("po")}")
print(f"File name ends with son: {"body.json".endswith("son")}")
```

Split and join

```python
path = '/usr/local/bin'
print(f"Path parts: {path.split("/")}")
print(f"Joined path: {"/".join(["/usr","local","bin"])}")
print("/usr"+"/local"+"/bin")
```

Access character & substring

```python
full_name = "Bershin John"
print(full_name[0])
print(full_name[-1])
print(full_name[4])
print(full_name[8:])
print(full_name[:7])
print(full_name[3:7])
```

Length of string

```python
full_name = "Bershin John"
print(len(full_name))
print(len("Bershin"))
```

String are immutable

```python
full_name = "Bershin John"
full_name[4] = 'd'
```
