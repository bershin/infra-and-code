### What is functions
Functions package reusable code into named blocks, improving modularity, readability and testability. They prevent duplication(DRY) and makes scripts easier to maintain.

```python
def greet_user(name):
    """Greets the user by name
    Args:
        name(str): The user to greet.
    Returns:
        None: Functions does not return any value.
    """
    print(f"Hello {name}!")

greet_user('Alice')
help(greet_user)
```

```python
import random
def generate_random(min_num, max_num):
    """Generate an integer between min and max
    Args:
        min_num (int): The lowest boundary of the interval
        max_num (int): The highest boundary of the interval
    Return:
        int: Generated random number
    """
    return random.randint(min_num, max_num)

generater_number = generate_random(0, 10)
print(f"Generated number: {generater_number}")
```

### Parameter
Mentioned in function definition and the variable name NOT change.
eg) "min_num and max_num" in def generate_random(min_num, max_num):
### Arguments
Mentioned in the calling function and it changes.
eg) "0 and 10" in generate_random(0, 10)

### Positional Arguments
Matches by order
### Keyword Arguments
Matches by name and can be out of order.
- If its a mix, positional argments should come before keyword arguments.

```python
def check_service_status(service_name, service_status):
    print(f"{service_name} is {service_status}")
    return True

check_service_status("nginx", "running")
check_service_status("running", "nginx")
check_service_status(service_status="running", service_name="nginx")
# check_service_status(service_status="running", "nginx") # Not allowed
# check_service_status("running", service_name="nginx") # Assign twice to service_name
check_service_status("running", service_status="nginx")
```

```python
def connect(host, port=22, timeout=30):
    print(f"Connect to host {host} on port {port} (timeout {timeout})")
connect("web01")
connect("web01", 443, 60)
connect("web01", 60) # Positional argument interfere
connect("web01", timeout=60)
```

```python
import socket

def check_port(host, port, timeout=5):
    """Check if a tcp port is open on a given host

    This check the TCP port for the host till the timeout.
    
    Args:
        host (str): Hostname or IPaddress.
        port (int): TCP port number.
        timeout (int, optional): Connection timeout in seconds. Default to 5.
    
    Return:
        bool: True if the port is open. False otherwise.
    """
    try:
        with socket.create_connection((host,port),timeout):
            return True
    except Exception:
        return False
    
print(check_port("www.google.com", 443))
print(check_port("www.google.com", 22))
print(check_port("www.sdfghasd.com", 22))
help(check_port)
```

```python
def greet_users(user_list):
    """Greet list of user by name.

    Args:
        user_list (list(str)): List of user to greet
    Return:
        None: 

    """
    for user in user_list:
        print(f"Hello, {user}!")

greet_users(['Alice', 'Bob', 'Charlie'])
```

```python
def sum_even(numbers):
    """ Calculate the sum of even numbers from a list.

        Args:
            numbers (list(int)): List of numbers to evaluate
        Return:
            int: Sum of even numbers
    """
    # Used generator instead of list(both works here)
    return sum(num for num in numbers if num % 2 == 0)

total_even = sum_even([1,2,3,4,5,6])
print(total_even)
```

```python
def fibonacci(n):
    """ Calculate the first n fibonnaci number

    Args:
        n (int): number of sequence

    Returns:
        list(int): Return a list of fibonnaci series
    """
    seq = [0, 1]
    for _ in range(2, n):
        seq.append(seq[-1] + seq[-2])
    
    return seq[:n]

print(fibonacci(10))
print(fibonacci(1))
```
