### List of sequencial number Vs Range
- Creating large lists for loops is memory-intensive(eg, list[range[1_000_000]])
- range() stores only start, stop & step values not all numbers.
    - Numbers are generated one at a time during iteration, reducing memory usage.
    - Ideal for loops needing a fixed number of iteration without large allocations.

```python
import sys

number_count = 1_000_000
number_list = list(range(number_count))
print(number_list[:20])

number_range = range(number_count)
print(number_range[:20])
for num in number_range[:20]:
    print(num)
```

```python
list_size_mb = sys.getsizeof(number_list) / 1024**2
print(f"list size: {list_size_mb:.2f}")
range_size_mb = sys.getsizeof(number_range) / 1024**2
print(f"Range size: {range_size_mb:.7f}")

print(f"List uses {(list_size_mb/range_size_mb):.2f} times more memory")
```
