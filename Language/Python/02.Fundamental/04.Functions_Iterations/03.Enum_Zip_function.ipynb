### Getting index + value: enumerate()
- Use enumerate(iterable, start=0) to get(index, item) pair
- The start parameter sets the initial index value.

```python
servers = ["web01", "web02", "web03"]
for server in enumerate(servers):
    print(server, type(server))
```

```python
servers = ["web01", "web02", "web03"]
for id, server in enumerate(servers, 1):
    print(f"server: 192.168.0.0{id}")
```

### Parallel iteration: zip()
- Use zip(*iteration) to pair item from multiple iterables.
- Iteration stops when the shortest iterable is exhausted.

```python
hosts = ["host-a", "host-b", "host-c"]
ips = ["10.0.0.2", "10.0.0.3"]
azs = ["us-east1", "us-west2"]

for host, ip, az in zip(hosts, ips, azs):
    print(f"host: {host}, ip: {ip}, az: {az}")
```
